```python
import pandas as pd
import plotly.express as px
import os
from google.colab import files
uploaded = files.upload()
df = pd.read_csv("all_farms_combined.csv")
df.head()
```

```python
def create_farm_map(df):

    color_map = {
        "Poultry": "#D3A628",
        "Swine": "#5C246E",
        "Cattle": "#A63A50",
        "Packaging": "#03FC17",
        "Spray Drying": "#42f5e9",
        "Other": "#f50000"
    }
    df["Revenue"] = df["Revenue"].apply(lambda x: f"${x:.1f}M" if pd.notnull(x) else "")
    fig = px.scatter_mapbox(df,
                             lat='Latitude',
                             lon='Longitude',
                             color='Type',
                             size=[11] * len(df),
                             size_max=8,
                             color_discrete_map=color_map,
                             hover_name='Name',
                             hover_data={
                                 'City': True,
                                 'State': True,
                                 'Revenue': True,
                                 'Latitude': False,
                                 'Longitude': False,
                                 'Type': True,
                                 'Info': False
                             },
                             zoom=5,
                             center={'lat': 41.5, 'lon': -89},
                             title='Production in Iowa, Illinois, and Indiana')


    fig.update_layout(
        mapbox_style="carto-positron",
        margin={"r":0,"t":50,"l":0,"b":0},
        height=600,
        width=800,
        title_x=0.5,
        title_font_size=16
    )

    return fig
def main():
    df = pd.read_csv("all_farms_combined.csv")
    output_file = "index.html"
    fig = create_farm_map(df)
    fig.write_html(output_file)
    abs_path = os.path.abspath(output_file)
    #print(f"Map saved to: {abs_path}")  - add this line if you can't find the map
    fig.show()

if __name__ == "__main__":
    main()
```
